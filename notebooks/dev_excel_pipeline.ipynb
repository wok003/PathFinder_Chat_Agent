### what could actually go wrong with a raw Excel file?
* Before code, list the failure modes, same as always:

* Multiple sheets — pandas defaults to reading only the first sheet; if the answer lives on a different one, you get nothing
* Header row isn't row 0 — some files have titles/merged cells above the real header, so column names come out garbled or as Unnamed: 0
* Extra formatting noise — currency symbols, commas in numbers, merged cells, totals rows mixed into data rows
* Wrong dtypes — a "sales" column read as string instead of numeric because of a stray $ or ,
