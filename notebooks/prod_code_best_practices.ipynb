By wrapping the problematic code within try block and specifying the exception type to catch 
in the except block, we can gracefully handle the error and prevent it from crashing the program.

try-except + else: else block executes code only if no excpetions are raised.

````python
try:
    result = int(input("Enter a number: "))
except ValueError:
    print("Invalid input. Please enter a valid integer.")
else:
    print("You entered:", result)
````

````python
try:
    denom = int(input("Enter the denominator: "))
    result = 10/denom # attempting division by zero

except (ValueError, ZeroDivisionError) as e:
    print("Error:", e)
````

finally block to execute cleanup code, regardless of whether an exception occurs or not.

````python
try:
    file = open("example.txt", "r")
    content = file.read()

except FileNotFoundError:
    print("File not found. error")

else:
    print("File content", content)

finally:
    if file:
        file.close() # Ensuring file is closed to ensure proper resource management
````

### Key Features of PyTest

* automatically discovers tests by looking for files that start with "test_" or end with "_test.py"

# Pytest Quick Reference Guide

A concise guide to testing classes, methods, and managing test organization in Python.

## 1. Project Structure & Organization
Pytest automatically discovers tests in files starting with `test_` and functions starting with `test_`.

**Recommended Layout:**
```text
project/
├── src/
│   └── models.py       # Your actual code

    ├── __init__.py
    ├── conftest.py     # Shared fixtures for the directory
    ├── test_models.py  # Tests for models.py
    └── test_utils.py   # Tests for utils.py

## 2. Basic Testing (State & Behavior)
Use `assert` to verify that methods return the correct values and attributes are set correctly.

```python
def test_attribute_initialization():
    obj = MyClass(name="Test")
    assert obj.name == "Test"  # Check Attribute

def test_method_behavior():
    obj = MyClass()
    result = obj.do_something()
    assert result == True       # Check Behavior
```

---

## 3. Handling Exceptions
Use `pytest.raises` to ensure your code fails correctly when it encounters invalid input.

```python
import pytest

def test_invalid_input_raises_error():
    with pytest.raises(ValueError, match="Invalid input"):
        obj.method_with_invalid_input("-1")
```

---

## 4. Fixtures (The "Setup" Logic)
Fixtures provide a way to "prepare" objects before tests run. They are injected into test functions as arguments.

```python
import pytest

@pytest.fixture
def sample_user():
    """Create a shared user object for multiple tests."""
    return User(username="test_user", id=1)

def test_user_login(sample_user):
    # 'sample_user' is automatically provided by the fixture
    assert sample_user.is_active is True
```

---

## 5. Fixture Scopes (Namespace/Lifetime)
Scopes define how often a fixture is reused, effectively managing the "scope" of the data.

| Scope | Description |
| :--- | :--- |
| `function` | (Default) Created anew for every test function. |
| `class` | Created once per test class. |
| `module` | Created once per `.py` file. |
| `session` | Created once for the entire test run (best for DB connections). |

**Example:**
```python
@pytest.fixture(scope="module")
def db_connection():
    # Connect once for all tests in this file
    return connect_to_db()
```

---

## 6. Parametrization (Data-Driven Testing)
Run the same test logic with multiple different inputs.

```python
import pytest

@pytest.mark.parametrize("input_val, expected_output", [
    (10, 20),
    (20, 40),
    (30, 60),
])
def test_multiplication(input_val, expected_output):
    assert input_val * 2 == expected_output
```

---

## 7. Shared Configuration (conftest.py)
Use `conftest.py` to define fixtures that are shared across multiple files in a directory without needing to import them.

**File: `tests/conftest.py`**
```python
@pytest.fixture
def shared_config():
    return {"api_key": "12345", "timeout": 30}
```
*(Now any test in `test_models.py` or `test_api.py` can use `shared_config` as a parameter).*

---

## 8. Test Marking (Filtering)
"Name" your tests or groups to select which ones to run from the CLI.

```python
@pytest.mark.smoke
def test_critical_path():
    pass

@pytest.mark.slow
def test_heavy_computation():
    pass
```

**CLI Commands:**
- `pytest` : Run all tests.
- `pytest -m smoke` : Run only tests marked as "smoke".
- `pytest -m "not slow"` : Run all tests except the "slow" ones.

Using Pytest Fixtures for Parameterized Tests
Pytest fixtures are powerful functions that help you write better, more maintainable tests by:
* **Writing parameterized test methods** where you can pass arguments to test functions. 
* **Reusing setup or context code** across multiple test methods avoiding repetition. 
* **managing test dependencies** and creating modular, composable test setups.
* **Handling setup and teardown** logic automatically


Fixtures are defined using the @pytest.fixture decorator placed above functions that provide setup data, test context, or resources. 

````python
%%writefile test_mylibrary.py
# PyTest test files typically start with "test_"
import pytest 
import pandas as pd 
import logging 

##############################
# Function under test
##############################
def import_data(pth):
    """Load CSV data into a Pandas DataFrame."""
    df = pd.read_csv(pth)
    return df 

##############################
# Fixture Definition
##############################

@pytest.fixture(scope="module")
def path():
    """ 
    Fixture that provides the path to the data file.
    The test function test_import_data() accepts this fixture as a parameter.
    """
    return "../baseline_metrics.csv"

#################################
# Test Method using the Fixture
#################################

def test_import_data(path):
    """Test the import_data function using the path fixture."""
    try: 
        df = import_data(path)
    except FileNotFoundError as err: 
        logging.error("File not found at path: %s", path)
        raise err 

    # validate the DataFrame Shape
    assert df.shape[0] > 0, "DataFrame has no rows"
    assert df.shape[1] > 0, "DataFrame has no columns"
````

### 2. Fixture Scopes
Fixtures can have differenr scopes that determine when they are created and destroyed:

**Scope** ----------------------                     **Description**
1.   function                       Run once per test function (serve fresh meal to every guest)
2.   class                          Run once per test class (once pot cooked, not it will be shared among everyone on the table)
3.   module                         Run once per test module (file) (pot cooked once, everyone in the resturant shares among each other.)
4.   package                        Run once per test package (everyone in the street also share once cooked)
5.   session                        Run once per test session (entire pytest run) (once cooked everyomne in the city shares)

## **LOGGING**

> **Logging helps understand events that occur while running your program.**
* say your model run overnight and the results following morning not what you expect.
* log messages can help you understand more about the context in those results occured.
* AIM IN THE FOLLOWING EXERCISE: learn about **qualities** that make a log message effective.

**Logging : recording messages which are describing events that have occured while running your software.**


- **tip1**: be professional and clear
- **tip2**: be concise and use normal capitalisation
- **tip3**: choose apt level for logging
    - **debug**: anything that happens in the program
    - **error**: record any error that occurs
    - **info**: user driven or system specific actions etc

- **tip4**: provide any useful information

### SCOPE:

* Any time your code leaves its "safe zone" and interacts with the outside world, you must log
* Whenever you use a try-except block, you must log inside the except block.
* Log when the "Status" of a process changes. If a process has a beginning, a middle, and an end, you should log the transitions.
* Log when something happens that isn't an "Error" but is "Suspicious." 
* Never log just a message; always log the data that makes the message unique. 

### Setting up logging in Python

````python
import logging

logging.basicConfig(
    filename='./results.log',
    level=logging.INFO,
    filemode='w',
    format='%(name)s - %(levelname)s - %(message)s'
)
````

Common Logging levels. (level name: severity value: When to Use)
> * DEBUG: 10: Detailed info for diagnosing problems. (Muttering to yourself while debugging)
> * INFO: 20: Confirmation things are working as expected. (Normal Status Updates)
> * WARNING: 30: Something unexpected but not breaking anything yet.(a yellow warning light)
> * ERROR: 40: something failed. (red light)
> * CRITICAL: 50: Serious failure- app may not continue. (fire alarm)

* Logging is crucial tool for **understanding** and **troubleshooting** the behavior of your applications.
