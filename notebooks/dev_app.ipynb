````python
from langchain_tavily import TavilySearch
from faster_whisper import WhisperModel

TAVILY_API_KEY='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'

class Toolbox:

    # tools access: web search
    def web_search(self, query: str) -> str:
        """Searches query/facts on the web and returns the top 3 pages with content

        Args:
            query: unknown fact. 
        """
        tavily_search = TavilySearch(max_results=3, tavily_api_key=TAVILY_API_KEY)
        data = tavily_search.invoke({"query": query})
        search_docs = data.get("results", data)
        return search_docs

    def fetch_page(self, url: str) -> str:
        """Fetch and extract the main text content of a webpage. Use this after
        web_search when you need details not visible in the search snippet.
        
        Args:
            url: web address of the webpage.
        """

        try:
            resp = requests.get(url, timeout=15, headers={"User-Agent": "Mozilla/5.0"})
            resp.raise_for_status()  # converts bad web page excpetion into error: 404, 503 etc.
        except requests.exceptions.Timeout:
            return f"Error: request to {url} timed out after 15 seconds."
        except requests.exceptions.ConnectionError:
            return f"Error: could not connect to {url} (site may be down or unreachable)."
        except requests.exceptions.HTTPError as e:
            return f"Error: {url} returned HTTP {resp.status_code}."
        except requests.exceptions.RequestException as e:
            return f"Error: failed to fetch {url} ({e})."

        soup = BeautifulSoup(resp.text, "html.parser")
        for tag in soup(["script", "style", "nav", "footer"]):
            tag.decompose()
        text = soup.get_text(separator="\n", strip=True)
        return text[:2000]  # cap length so it doesn't blow your context window
````

````python
from langgraph.graph import StateGraph, START, END
from langgraph.prebuilt import ToolNode, tools_condition
from IPython.display import Image, display
from langgraph.graph import MessagesState

class Workflow_builder:

    def __init__(self, agent_framework):
        self.agent_framework = agent_framework
        self.react_workflow = None

        print("Workflow building started...")
        self.workflow()
        print("Workflow built successfully!!!")

    # create the stateGraph
    def workflow(self):
        builder = StateGraph(MessagesState)
        builder.add_node("node_1", self.node_assistant)
        builder.add_node("tools", ToolNode(self.agent_framework.tools_list))
        builder.add_node("formatter", self.format_node)

        # add edges 
        builder.add_edge(START, "node_1")
        builder.add_conditional_edges("node_1", tools_condition,
                                      {
                                          "tools": "tools",
                                          "__end__": "formatter"
                                      })
        builder.add_edge("tools", "node_1")
        builder.add_edge("formatter", END)

        # compile the workflow with the memory
        self.react_workflow = builder.compile(checkpointer=self.agent_framework.memory)

        # 
        print("built node")
        self.visualise_workflow()

    # define node working
    def node_assistant(self, State: MessagesState):
        # do something with the input state
        print("node_assistant running")
        sys_msg = SystemMessage(content= REASONER_PROMPT)        
        # Stage 1: REASONING/TOOL-USE AGENT
        response = self.agent_framework.llm_with_tools.invoke([sys_msg]+ 
                                                               State["messages"]
                                                               )
        print("node assistant finished")                                                
        return {"messages": [response]}

    def format_node(self, state: MessagesState):
        print("format node running")
        prompt = state["messages"][0].content
        last_answer = "Final Answer:" +state["messages"][-1].content
        print(f"original question: {prompt}, {FORMATTER_PROMPT}, {last_answer}")

        response = self.agent_framework.llm.invoke([
            SystemMessage(content=prompt),
            HumanMessage(content=last_answer),
            SystemMessage(content=FORMATTER_PROMPT)
        ])
        print("format node finished")
        return {"messages": [response]}

    def format_answer(self, raw_output: str) -> str:
        import re 
        import json
        match = re.search(r'\[.*\]', raw_output, re.DOTALL)
        if not match:
            return raw_output.strip()
        try:
            items = json.loads(match.group())
            return ", ".join(str(item).strip() for item in items)  # comma + space, per your rule
        except json.JSONDecodeError:
            return raw_output.strip()

    def visualise_workflow(self):
        display(Image(self.react_workflow.get_graph().draw_mermaid_png()))
        
````

````python
from langchain_core.messages import SystemMessage

class BasicAgent:
   
    def __init__(self):
        print("BasicAgent initialized.")

        # 1. Create the agent object here   
        self.agent_framework = AgentBuilder()

        # 2. Create the workflow object here
        self.workflow = Workflow_builder(self.agent_framework)

        # 3. Initialise a thread_id variable
        self.thread_id = 1

    def __call__(self, question: str) -> str:

        print(f"Agent received question (first 50 chars): {question[:50]}...")
        fixed_answer = "This is a default answer."
        print(f"Agent returning fixed answer: {fixed_answer}")

        answer = self.query_feeder(question)

        
        #final_response = answer['messages'][-1].content[0]['text'] # use: google flash model
        final_response = answer
        print(f"Agent returning answer: {answer}")
        self.thread_id = self.thread_id + 1
        return final_response

    def query_feeder(self, question: str) -> str:
        # Invoke logic
        # ---------------- Agent Ready for Basic Question Answering ------------------------------------------
        # change the thread_id every time
        self.agent_framework.config["configurable"]["thread_id"] = str(self.thread_id)    
        return self.workflow.react_workflow.invoke({"messages": [question]}, self.agent_framework.config)
````
